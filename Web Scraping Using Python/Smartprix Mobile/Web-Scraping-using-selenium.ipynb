```python
import pandas as pd
import numpy as np
import time

from bs4 import BeautifulSoup
```

```python
## Loading HTML file Here

with open('smartprix.html', 'r', encoding='utf-8') as f:
    html = f.read()
```

```python
soup = BeautifulSoup(html, 'lxml')
```

```python
# sm-product has-tag has-features has-actions

product = soup.find_all('div', class_='sm-product has-tag has-features has-actions')
len(product)
```

```python
model = []
price = []
star = []
rating = []
display = []
camera = []
card = []
os = []
```

```python
for i in product:
    model.append(i.find('h2', class_='sm-product-title').text.strip())
    price.append(i.find('span', class_='price').text.strip())
    star.append(i.find('span', class_='rating').text.strip())
    rating.append(i.find('span', class_='total-rating').text.strip())
```

```python
for i in product:
    title = i.find('h2', class_='sm-product-title')
    model.append(title.get_text(strip=True) if title else None)

    price_ = i.find('span', class_='price')
    price.append(price_.get_text(strip=True) if price_ else None)

    star_ = i.find('span', class_='rating')
    star.append(star_.get_text(strip=True) if star_ else None)

    rating_ = i.find('span', class_='total-rating')
    rating.append(rating_.get_text(strip=True) if rating_ else None)
```

```python
def safe_text(tag):
    return tag.get_text(strip=True) if tag else None

for i in product:
    display.append(safe_text(i.find('li', class_='highlight display')))
    camera.append(safe_text(i.find('li', class_='highlight camera')))
    card.append(safe_text(i.find('li', class_='highlight card')))
    os.append(safe_text(i.find('li', class_='highlight os')))
```

```python
len(display)
len(camera)
len(card)
len(os)
```

```python
# Making DataFrame or Dataset

smartprix_smartphone = pd.DataFrame({
    'model': model,
    'price': price,
    'star': star,
    'rating': rating,
    'display': display,
    'camera': camera,
    'card': card,
    'os': os
})
```
