```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

# Why Seaborn ?

### Types of Function 

# 1> Relation Plot

```python
tips = sns.load_dataset('tips')
tips
```

```python
gap = px.data.gapminder()
```

```python
gap
```

### i> Scatter Plot

```python
# Axes level function

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
# Figure level function

sns.relplot(data=tips, x='total_bill', y='tip', hue='sex', size='size', style='time', kind='scatter')
```

### ii> Line plot

```python
df = gap[gap['country']=='India']
df
```

```python
# Axes level function

sns.lineplot(data=df, x='year', y='lifeExp')
```

```python
# figure level

sns.relplot(data=df, x='year', y='lifeExp',kind='line')
```

```python
# hue  ---> Style

temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
temp_df
```

```python
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country')
```

```python
temp_df1 = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df1
```

```python
sns.relplot(data=temp_df1, kind='line', x='year', y='lifeExp', hue='country', style='continent')
```

```python
# Facet Plot  ----> Figure level function -->work with relplot

sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', col='sex')
```

```python
sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', row='sex')
```

```python
sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', col='smoker', row='sex')
```

```python
sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', col='sex', row='day')
```

```python
gap.head(1)
```

```python
# Col wrap

sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```

# 3> Matrix Plot

### i> Heatmap 

```python
# Plot rectangular data as a color encoded matrix
```

```python
temp_df = gap.pivot(index='country', columns='year', values='lifeExp')
```

```python
# Axes level function
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot
```

```python
temp_df1 = gap[gap['continent']=='Europe'].pivot(index='country', columns='year', values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp_df1, annot=True, linewidth=0.5, cmap='viridis')
```
