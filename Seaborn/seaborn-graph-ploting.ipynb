# Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Seborn:
       Seaborn is a statistical data visualization library built on top of Matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics with minimal code.
```

```python
Key Features:

1> Statistical visualization focus
2> Beautiful default styles and color palettes
3> Built-in statistical functions (regression, distribution fitting)
4> Easy handling of categorical data
5> Integration with pandas DataFrames
6> Advanced plot types for statistical analysis
```

```python
Typical Use Cases:

1> Statistical analysis and modeling
2> Exploratory data analysis (EDA)
3> Research and academic presentations
4> Correlation and relationship analysis
5> Distribution analysis
```

## Graph Types and Examples

### 1. Distribution Plots:

```python
1. Distribution Plots:

Histogram with KDE

Description: Enhanced histograms with kernel density estimation overlay.
Use Case: Distribution analysis with smooth density curves.
```

```python
np.random.seed(42)
tips = sns.load_dataset('tips')

plt.figure(figsize=(10, 6))
sns.histplot(data=tips, x='total_bill', kde=True, bins=20)
plt.title('Distribution of Total Bill with KDE')
plt.show()

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(data=tips, x='total_bill', hue='sex', kde=True, alpha=0.7)
plt.title('Total Bill Distribution by Gender')

plt.subplot(1, 2, 2)
sns.histplot(data=tips, x='tip', hue='time', kde=True, alpha=0.7)
plt.title('Tip Distribution by Time')
plt.tight_layout()
plt.show()
```

### 2. Box Plot and Violin Plot

```python
Box Plot and Violin Plot:

Description: Enhanced box plots and violin plots for distribution comparison.

Use Case: Comparing distributions across categories with detailed shape information.
```

```python
# 2> Box Plot and Violin Plot

# Box plot
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill by Day of Week')

# Violin plot
plt.subplot(1, 2, 2)
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex')
plt.title('Total Bill Distribution by Day and Gender')
plt.tight_layout()
plt.show()
```

### 3.Relationship Plots

```python
Scatter Plot with Regression:

Description: Scatter plots with automatic regression lines and confidence intervals.

Use Case: Correlation analysis with statistical modeling.
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', size='size')
sns.regplot(data=tips, x='total_bill', y='tip', scatter=False, color='red')
plt.title('Relationship between Total Bill and Tip')
plt.show()

numeric_columns = ['total_bill', 'tip', 'size']
sns.pairplot(data=tips[numeric_columns + ['sex']], hue='sex', diag_kind='kde')
plt.suptitle('Pairwise Relationships in Tips Dataset', y=1.02)
plt.show()
```

### 4.Correlation Heatmap

```python
Correlation Heatmap:

Description: Correlation matrix visualization with hierarchical clustering.

Use Case: Feature correlation analysis, multicollinearity detection.
```

```python
iris = sns.load_dataset('iris')
corr_matrix = iris.select_dtypes(include=[np.number]).corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0,
           square=True, linewidths=0.5)
plt.title('Correlation Matrix of Iris Dataset')
plt.show()

plt.figure(figsize=(8, 6))
sns.clustermap(corr_matrix, annot=True, cmap='coolwarm', center=0)
plt.show()
```

### 5. Matrix Plots

```python
 Matrix Plots

Description: 2D data representation using color intensity.

Use Case: Pivot table visualization, time series patterns, confusion matrices.
```

```python
flights = sns.load_dataset('flights')
flight_pivot = flights.pivot(index='month', columns='year', values='passengers')

plt.figure(figsize=(12, 8))
sns.heatmap(flight_pivot, annot=True, fmt='d', cmap='YlOrRd')
plt.title('Flight Passengers by Month and Year')
plt.show()
```

### 6. Regression Plots

```python
Regression Plots
Advanced Regression Analysis

Description: Multiple types of regression with residual analysis.
Use Case: Model fitting, prediction analysis, residual diagnostics.
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.regplot(data=tips, x='total_bill', y='tip', ax=axes[0,0])
axes[0,0].set_title('Linear Regression: Total Bill vs Tip')

sns.residplot(data=tips, x='total_bill', y='tip', ax=axes[0,1])
axes[0,1].set_title('Residuals vs Total Bill')

sns.regplot(data=tips, x='total_bill', y='tip', order=2, ax=axes[1,0])
axes[1,0].set_title('Polynomial Regression (degree=2)')

sns.regplot(data=tips, x='size', y='total_bill', ax=axes[1,1])
axes[1,1].set_title('Party Size vs Total Bill')

plt.tight_layout()
plt.show()
```

### 7. Statistical Plots

```python
Statistical Plots:
Distribution Comparison
Description: Compare distributions across categories using various statistical plots.
Use Case: A/B testing, group comparisons, statistical hypothesis testing.
```

```python

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
sns.stripplot(data=tips, x='day', y='total_bill', ax=axes[0,0])
axes[0,0].set_title('Strip Plot: Total Bill by Day')

sns.swarmplot(data=tips, x='day', y='total_bill', ax=axes[0,1])
axes[0,1].set_title('Swarm Plot: Total Bill by Day')

sns.boxplot(data=tips, x='day', y='total_bill', ax=axes[1,0])
sns.stripplot(data=tips, x='day', y='total_bill', ax=axes[1,0], 
              color='red', alpha=0.5, size=3)
axes[1,0].set_title('Box Plot with Individual Points')

sns.violinplot(data=tips, x='day', y='total_bill', inner='quartile', ax=axes[1,1])
axes[1,1].set_title('Violin Plot with Quartiles')

plt.tight_layout()
plt.show()
```
