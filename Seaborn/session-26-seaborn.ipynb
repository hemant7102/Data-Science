```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
import numpy as np
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

# 4> Catrgorical Plots

### 1> Categorical Scatter Plots

#### Strip Plot

```python
# Strip Plot
# Axes Level Function

sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
# Strip Plot
# Axes Level Function
# Jitter
sns.stripplot(data=tips, x='day', y='total_bill', jitter=False)
```

```python
# Strip Plot
# Axes Level Function
# Jitter
sns.stripplot(data=tips, x='day', y='total_bill', jitter=0.2)
```

```python
# Catplot Plot
# Figure Level Function
# Jitter
sns.catplot(data=tips, x='day', y='total_bill', jitter=0.2, kind='strip')
```

#### Swarm Plot

```python
# Axes level function

sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex')
```

```python
# Figure Level function

sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
```

### 2> Cateorical Distribution Plot

#### Box Plot

```python
# Axes Level function

sns.boxplot(data=tips, x='sex', y='total_bill')
```

```python
# Figure Level Function

sns.catplot(data=tips, x='day', y='total_bill', hue='sex')
```

```python
# Single BoxPlot  -----> Numrical Col

sns.boxplot(data=tips, y='total_bill')
```

#### Violin Plot = (Box Plot + KDE Plot)

```python
# Axes Level Function

sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
# Figure level Function

sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex')
```

```python
# Figure level Function

sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex', split=True)
```

### 3> Categorical Estimate Plot ---> for central tendency

```python
# ci=None
```

#### Bar Plot

```python
# Axes level function

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker')
```

```python
# Axes level function
# estimator by default is mean

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.std)
```

```python
# Axes level function
# estimator by default is mean

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.median, ci=None)
```

```python
# Axes level function
# estimator by default is mean

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.min)
```

```python
# Axes level function
# estimator by default is mean

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.max)
```

#### Point Plot

```python
sns.pointplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.min, ci=None)
```

#### Count Plot

```python
sns.countplot(data=tips, x='sex', hue='day')
```

```python
# Faceting Using Catplot

sns.catplot(data=tips, x='sex', y='total_bill', col='smoker', kind='box', row='time')
```

# 6> MultiGrid OR MultiPlot

### 1> FacetGrid ---> A second way to plot facet plot

```python
# Figure Level Function   ---> Relplot ---> displot  ---> catplot ---> lmplot
```

```python
g = sns.FacetGrid(data=tips, col='day', row='time', hue='smoker')
g.map(sns.violinplot, 'sex', 'total_bill')
g.add_legend()
```

### Plotting Pairwise Relationship (PairGrid Vs PairPlot)

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris, hue='species')
```

### PairGird

```python
g = sns.PairGrid(iris, hue='species')
g.map(sns.scatterplot)
```

```python
# Map_diag  ---> map_offdiag

g = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_diag ---> map_upper   ---> map_lower

g = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

### JointGrid  Vs JointPlot

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='hist', hue='sex')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='kde', hue='sex')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex')
```

#### JointGrid

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.histplot)
```

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.boxplot)
```

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.violinplot)
```
