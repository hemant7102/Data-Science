# More Methods On GroupBY

```python
# agg method
# passing dict

genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# passing list

genres.agg(['min','max','mean'])
```

```python
# Adding Both Syntax

genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# Looping on groups

for group,data in genres:
    print(type(group),type(data))
    print(data)
```

```python
# find the highest rated movie of each genre.

df = pd.DataFrame(columns=movies.columns)
print(df)

for group,data in genres:
    df = df.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])


df    
```

## Examples

```python
def foo(group):
    print(group)
    return group
    
genres.apply(foo)
```

```python
# 1> find number of movies starting with 'A' for each genre group.

def start(group):
    return group['Series_Title'].str.startwith('A').sum()

   
```

```python
genres.apply(start) 
```

```python
# find ranking of each movies in the group according to IMDB Rating

def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
    return group

genres.apply(rank_movie)    
```

```python
# find normalized IMDb group wise

def normal(group):
    group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group

genres.apply(normal)
```

## GroupBy on Multiple Columns

```python
# size
# get_group

duo = movies.groupby(['Director', 'Star1'])
duo.size()
```

```python
# find the most earning actor ---> director combo

duo['Gross'].sum().sort_values(ascending=False).head()
```

```python
# find the best in terms of metascore avg actor --> genre

movies.groupby(['Star1','Genre'])['Metascore'].mean().reset_index().sort_values('Metascore',ascending=False)
```
