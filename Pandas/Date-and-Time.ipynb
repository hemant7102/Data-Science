# Date-Range

# 1> Genratebdialy dates in a given range

```python
# All date 

pd.date_range(start='2025/10/1', end='2025/10/30', freq='D')
```

```python
# Alternates Dates 

pd.date_range(start='2025/10/1', end='2025/12/30', freq='2D')
```

```python
print(4 + 3 % 5)
```

```python
# 3 days gap

pd.date_range(start='2025/10/1', end='2025/10/30', freq='3D')
```

```python
pd.date_range(start='2025/10/1', end='2025/12/1', freq='B')
```

```python
# freq='W-Thus' or 'W-Wed'

pd.date_range(start='2025/10/1', end='2025/10/30', freq='W-THU')
```

# to_datetime Function

### With Error

```python
s = pd.Series(['2025/10/1', '2026/10/1', '2027/10/1'])
s
```

```python
# To convert existing object into pandas Timestamp and DatetimeIndex object.

pd.to_datetime(s)
```

```python
pd.to_datetime(s, errors='coerce').dt.year
```

```python
pd.to_datetime(s, errors='coerce').dt.month
```

```python
pd.to_datetime(s, errors='coerce').dt.day
```

```python
pd.to_datetime(s, errors='coerce').dt.month_name()
```

```python
pd.to_datetime(s, errors='coerce').dt.day_name()
```
