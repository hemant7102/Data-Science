# Pandas

```python
 Pandas:
    Pandas is primarily a data manipulation library, but it includes built-in plotting functionality that provides quick and convenient visualization directly from DataFrames and Series. It uses Matplotlib as its backend, making it perfect for rapid data exploration. 
```

```python
 Key Features:

 1> Direct plotting from DataFrames and Series
 2> Minimal code required for common plots
 3> Automatic handling of data types and indexing
 4> Integration with time series data
 5> Built on top of Matplotlib
```

```python
 Typical Use Cases:

 1> Quick data exploration and analysis
 2> Time series visualization
 3> Simple statistical plots during data cleaning
 4> Prototyping visualizations before creating publication-ready versions
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## Graph Types and Examples

```python
 1. Line Plot :

Description: Quick line plots directly from DataFrame columns, ideal for time series.

Use Case: Time series analysis, trend identification, multiple series comparison.
```

```python
# 1> Example of line plot

dates = pd.date_range('2023-01-01', periods=100, freq='D')
df = pd.DataFrame({
    'Date': dates,
    'Sales': np.cumsum(np.random.randn(100)) + 100,
    'Marketing': np.cumsum(np.random.randn(100)) + 80,
    'Customer_Acquisition': np.cumsum(np.random.randn(100)) + 50
})
df.set_index('Date', inplace=True)

# Simple line plot
df.plot(kind='line', figsize=(12, 6), title='Business Metrics Over Time')
plt.ylabel('Values')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)
plt.show()

df['Sales'].plot(kind='line', figsize=(10, 5), color='blue', linewidth=2,
                title='Sales Trend Analysis')
plt.ylabel('Sales ($)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
2. Bar Plot: 

Description: Horizontal and vertical bar charts for categorical data comparison.

Use Case: Category comparison, ranking analysis, survey results.
```

```python
# 2> Bar Plot Example

categories_df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Headphones'],
    'Q1_Sales': [45, 78, 23, 34, 56],
    'Q2_Sales': [52, 85, 28, 41, 63],
    'Q3_Sales': [48, 92, 31, 38, 59]
})
categories_df.set_index('Product', inplace=True)

# Vertical bar plot
categories_df.plot(kind='bar', figsize=(10, 6), 
                  title='Quarterly Sales by Product')
plt.ylabel('Sales (Units)')
plt.xticks(rotation=45)
plt.legend(title='Quarter')
plt.tight_layout()
plt.show()

# Horizontal bar plot
categories_df['Q3_Sales'].plot(kind='barh', figsize=(8, 6), 
                              color='skyblue',
                              title='Q3 Sales by Product')
plt.xlabel('Sales (Units)')
plt.show()

# Stacked bar chart
categories_df.plot(kind='bar', stacked=True, figsize=(10, 6),
                  title='Stacked Quarterly Sales')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
3. Histogram : 

Description: Distribution analysis of numerical columns with automatic binning.

Use Case: Understanding data distribution, identifying outliers, data quality assessment.
```

```python
# 3> Histogram Example

np.random.seed(42)
hist_df = pd.DataFrame({
    'Normal': np.random.normal(50, 15, 1000),
    'Exponential': np.random.exponential(2, 1000),
    'Uniform': np.random.uniform(0, 100, 1000)
})

hist_df['Normal'].plot(kind='hist', bins=30, alpha=0.7, 
                      figsize=(10, 6), color='lightblue',
                      title='Distribution of Normal Data')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.axvline(hist_df['Normal'].mean(), color='red', linestyle='--', 
           label=f'Mean: {hist_df["Normal"].mean():.1f}')
plt.legend()
plt.show()

hist_df.plot(kind='hist', bins=30, alpha=0.7, 
            figsize=(15, 5), subplots=True, layout=(1, 3),
            title='Distribution Comparison')
plt.tight_layout()
plt.show()
```

```python
4. Box Plot:

Description: Statistical summary visualization showing quartiles, median, and outliers.

Use Case: Comparing distributions across categories, outlier detection.
```

```python
# 4> Box Plot Example

box_data = pd.DataFrame({
    'Group_A': np.random.normal(100, 15, 100),
    'Group_B': np.random.normal(110, 20, 100),
    'Group_C': np.random.normal(95, 10, 100),
    'Group_D': np.random.normal(105, 25, 100)
})
box_data.plot(kind='box', figsize=(10, 6),
             title='Distribution Comparison Across Groups')
plt.ylabel('Values')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
5. Scatter Plot:

Description: Relationship visualization between two numerical columns.

Use Case: Correlation analysis, pattern identification, outlier detection.
```

```python
# 5> Scatter Plot Example

scatter_df = pd.DataFrame({
    'X': np.random.randn(200),
    'Y': 2 * np.random.randn(200) + np.random.randn(200),
    'Size': np.random.randint(20, 100, 200),
    'Category': np.random.choice(['A', 'B', 'C'], 200)
})
scatter_df['Y_corr'] = 1.5 * scatter_df['X'] + np.random.randn(200) * 0.5
scatter_df.plot(kind='scatter', x='X', y='Y_corr', 
               figsize=(8, 6), alpha=0.6,
               title='Scatter Plot: X vs Y with Correlation')
plt.grid(True, alpha=0.3)
plt.show()
scatter_df.plot(kind='scatter', x='X', y='Y_corr', s='Size',
               figsize=(8, 6), alpha=0.6,
               title='Scatter Plot with Variable Point Sizes')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
6. Area Plot:

Description: Filled area under lines, useful for showing cumulative data.

Use Case: Cumulative analysis, stacked categories, proportion visualization.
```

```python
# 6> Area Plot Example

area_df = pd.DataFrame({
    'Product_A': np.cumsum(np.random.randint(5, 15, 20)),
    'Product_B': np.cumsum(np.random.randint(3, 12, 20)),
    'Product_C': np.cumsum(np.random.randint(2, 10, 20))
})

area_df.plot(kind='area', figsize=(10, 6), alpha=0.7,
            title='Cumulative Sales by Product')
plt.ylabel('Cumulative Sales')
plt.xlabel('Time Period')
plt.legend(loc='upper left')
plt.show()
```

```python
7. Pie Chart:

Description: Proportion visualization using circular sectors.

Use Case: Market share analysis, budget breakdown, category proportions.
```

```python
# 7> Pie Chart Example

pie_data = pd.Series([30, 25, 20, 15, 10], 
                    index=['Product A', 'Product B', 'Product C', 'Product D', 'Others'],
                    name='Market Share')

pie_data.plot(kind='pie', figsize=(8, 8), autopct='%1.1f%%',
             title='Market Share Distribution')
plt.ylabel('')  
plt.show()
```

### Functionality and Plot Types

```python
1> Basic Plots: Line, bar, histogram, box, scatter, area, pie.
2> Customization: Limited - relies on matplotlib for advanced customization
3> Statistical Features: Basic statistical plots only
4> Time Series: Excellent built-in support for time-indexed data
```

### Data Integration

```python
1> DataFrame Integration: Native - plots directly from DataFrame columns
2> Index Handling: Automatic use of DataFrame index for x-axis
3> Data Types: Automatic handling of datetime, categorical data
4> Missing Data: Built-in handling of NaN values
```

### Aesthetics and Styling

```python
1> Default Style: Basic matplotlib defaults
2> Customization: Requires matplotlib code for styling
3> Color Handling: Basic color options
4> Themes: Limited built-in themes
```

### Performance

```python
1> Speed: Very fast for simple plots
2> Memory: Efficient - minimal overhead
3> Large Data: Good performance with large datasets
4> Streaming: Works well with chunked data processing
```

```python
Use Pandas Plotting When:

1> Quick data exploration during analysis
2> Simple visualizations of DataFrame contents
3> Time series plotting
4> Prototyping before creating publication plots
5> Working with small to medium datasets
6> Need minimal code for basic plots
```
