```python
import pandas as pd
import numpy as np
import datetime as dt
import matplotlib.pyplot as plt
```

## Creating Using Timestamp objects

### 1> Timedelta

```python
t1 = pd.Timestamp('6th Jan 2025 8:20:14')
t2 = pd.Timestamp('26th Jan 2025 10:00:00')

t2 - t1
```

## Standalone Creation

```python
pd.Timedelta(days=2, hours=10, minutes=35)
```

## Arithmetic Operation

```python
pd.Timestamp('2025/10/1') + pd.Timedelta(days=2, hours=10, minutes=35)
```

## Range using Timedelta

```python
pd.date_range(start='2025-10-1', end='2025-11-30', freq='D') + pd.Timedelta(days=2, hours=10, minutes=35)
```

```python
pd.date_range(start='2025-10-1', end='2025-11-30', freq='D') - pd.Timedelta(days=2, hours=10, minutes=35)
```

# How to handle dataset

### 1> Partial indexing --> select a particular year/month

```python
google.loc['2021-12']
```

```python
google.loc['2021']
```

### 2> Create year, month_name and day_name

```python
google['month_name'] = google.index.month_name()
```

```python
google.head()
```

```python
google['weekday_name'] = google.index.day_name()
```

```python
google.head()
```

```python
google['quarter'] = google.index.quarter
```

```python
google.head()
```

### 3> Slicing

```python
google.loc['2020-12-15':'2021-12-15']
```

```python
google.loc['2020-12-15':'2021-12-15':2]
```

### 4> Take challenge --> fetch data for every year 

```python
pd.date_range(start='2005-1-6', end='2022-1-6', freq=pd.DateOffset(years=1))
```

```python
google[google.index.isin(pd.date_range(start='2005-1-6', end='2021-1-6', freq=pd.DateOffset(years=1)))]
```

```python
google.head()
```

# Resampling

### 1> Downsampling

```python
google['Close'].plot()
```

```python
google['Close'].resample('M').mean().plot()   # month wise
```

```python
google['Close'].resample('Q').mean().plot()  # quarter wise
```

```python
google['Close'].resample('Y').mean().plot()   # Year wise
```

### 2> Upsampling

```python
google['Close'].resample('H').interpolate(method='linear').plot()
```
