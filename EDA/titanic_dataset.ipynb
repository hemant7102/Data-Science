### SibSp And Parch Columns

```python
df['family_size'] = df['SibSp'] + df['Parch'] + 1
```

```python
# 1 ---> alone
# 2 ---> small
# 3 ---> large
```

```python
def trasform_family_size(num):

    if num == 1:
        return "alone"

    elif num > 1 and num < 5:
        return "small"

    else:
        return "large"
```

```python
df['family_type'] = df['family_size'].apply(trasform_family_size)
```

```python
df
```

```python
pd.crosstab(df['Survived'], df['family_type'], normalize = 'columns')*100
```
