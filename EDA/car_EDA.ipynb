# Working with Columns

```python
# Changing the data type of columns

price_col = ['msrp', 'invoice_price', 'used/new_price']

for col in price_col:
    df[col] = df[col].replace('[\$,]','', regex=True).astype(float)
```

```python
df.info()
```

```python
# Calculate average difference
diff = (df['msrp'] - df['invoice_price']).mean()
```

```python
diff
```

```python
# Fill the missing value in Invoice Price Column

df['invoice_price'].fillna(df['msrp'] - diff, inplace=True)
```

```python
# cylinders

df['cylinders'] = df['cylinders'].astype(str).str.extract('(\d+)')
df['cylinders'] = pd.to_numeric(df['cylinders'], errors='coerce')

df['cylinders'] = df.groupby(
    ['make', 'body_style', 'engine_aspiration']
)['cylinders'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan)
)

df['cylinders'] = df.groupby(
    ['make', 'body_style']
)['cylinders'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan)
)


df['cylinders'] = df.groupby('make')['cylinders'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan)
)

df['cylinders'].fillna(df['cylinders'].mode()[0], inplace=True)
```

#### Horsepower col

```python
df['horsepower'] = df['horsepower'].astype(str).str.extract('(\d+)')
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
```

```python
df['horsepower'] = df.groupby(
    ['make', 'cylinders']
)['horsepower'].transform(
    lambda x: x.fillna(x.median())
)

df['horsepower'].fillna(df['horsepower'].median(), inplace=True)
```

#### torque col

```python
df['torque'] = df['torque'].astype(str).str.extract('(\d+)')
df['torque'] = pd.to_numeric(df['torque'], errors='coerce')
```

```python
df['torque'] = df.groupby(
    ['make', 'cylinders']
)['torque'].transform(
    lambda x: x.fillna(x.median())
)

df['torque'].fillna(df['torque'].median(), inplace=True)
```

#### Highway Fuel economy col

```python
df['highway_fuel_economy'] = df['highway_fuel_economy'].astype(str).str.extract('(\d+)')
df['highway_fuel_economy'] = pd.to_numeric(df['highway_fuel_economy'], errors='coerce')
```

```python
df['highway_fuel_economy'] = df.groupby(
    ['body_size', 'cylinders']
)['highway_fuel_economy'].transform(
    lambda x: x.fillna(x.median())
)

df['highway_fuel_economy'].fillna(df['highway_fuel_economy'].median(), inplace=True)
```

## Working with Outliers

```python
num_cols = ['invoice_price', 'horsepower', 'torque', 'highway_fuel_economy']

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(y=df[col])
    plt.title(f'Outliers in {col}')
    plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
Q1 = df['invoice_price'].quantile(0.25)
Q3 = df['invoice_price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

```python
outliers = df[(df['invoice_price'] < lower) | (df['invoice_price'] > upper)]
print("Number of outliers:", len(outliers))
```

```python
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower) | (df[column] > upper)]
    print(f"{column}: {len(outliers)} outliers")

for col in ['invoice_price', 'horsepower', 'torque', 'highway_fuel_economy']:
    detect_outliers(col)
```
