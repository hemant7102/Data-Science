```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('batsman_runs_ipl.csv')
df.head()
```

```python
runs = df.groupby('batter').sum()['batsman_run']
runs
```

```python
runs = runs.values
runs
```

```python
runs.shape
```

```python
sns.distplot(runs)
```

```python
m = 1000
n = 50

new = []

for i in range(m):
    x=np.random.choice(runs, size=n)
    new.append(x.mean())
```
