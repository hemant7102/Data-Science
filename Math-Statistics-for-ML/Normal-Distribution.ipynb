```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Standard Normal Varite(standard normal distribution)   (Z)

```python
df = pd.read_csv('train (1).csv')
```

```python
df.head()
```

```python
sns.kdeplot(df['Age'])
```

```python
df['Age'].mean()
```

```python
df['Age'].std()
```

```python
# find SNV 

x = (df['Age'] - df['Age'].mean()) / df['Age'].std()
```

```python
sns.kdeplot(x)
```

```python
x.mean()
```

```python
x.std()
```

```python
# Find the skew of titanic

df['Age'].skew()
```
