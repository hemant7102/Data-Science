# Linear Algebra - part 1

### How to calculate the magnituted ||A||  from distance

```python
# Distance from origin  ---> euclidean norm
# Define an n-dimensional vector A
A = np.array([1,2,3,4,5])

# Calculate the Euclidean distance from the origin (L2 norm)
distance = np.linalg.norm(A)

print("Euclidean distance from the origin : ", distance)
```

### Euclidean Distance

```python
# N-Dimension space

# To find two n-D vectors A and B
A = np.array([1,2,3,4,5])
B = np.array([6,7,8,9,10])

# Calculate the difference vectors
difference = A - B
difference
```

```python
# Calculate the Euclidean distance between A and B (L2 norm of the difference)

distance = np.linalg.norm(difference)
print("Euclidean Distance between A and B :", distance)
```

#### ML Algo

```python
# K-nearest-neighbors  ---> classification

# 1. Generate 5 random 3D Vectors
vector = np.random.rand(5,3)

# 2. Assign a random class (0 or 1) to each vector
classes = np.random.randint(0,2,5)

# 3. Plot the vector on a 3D Matplotlib graph
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
for vec, cls in zip(vector, classes):
    ax.quiver(0, 0, 0, vec[0], vec[1], vec[2], color=("r" if cls == 1 else "b"))

# 4. Get user input for a query point (3D vector)
query_vector = np.array(list(map(float, input("Enter the query point (3D Vector) separated by space :").split())))
# Input 1 1 1


# 5. Calculate the distance from the query vector to the 5 vector and find the nearest neighbor
distance = [euclidean(query_vector, vec) for vec in vector]
nearest_neighbor_index = np.argmin(distance)

# 6. Output the class of the nearest neighbor
print("The class of the nearest neighbor is : ", classes[nearest_neighbor_index])

# 7. Plot a query vectors with a different colors
ax.quiver(0, 0, 0, query_vector[0], query_vector[1], query_vector[2], color="g")
ax.set_xlabel("X_axis")
ax.set_ylabel("Y_axis")
ax.set_zlabel("Z_axis")
plt.show()
```

### Mean Centering

```python
# Without Mean Centering

# Generate random 2D data with 100 point
data = np.random.rand(100,2)
# print(data)

# Plot the original data
plt.figure(figsize = (12, 6))
plt.subplot(1, 2, 1)
plt.scatter(data[:, 0], data[:, 1], label = "original data")
plt.xlabel("X_axis")
plt.ylabel("Y_axis")
plt.title("Before Mean Centering")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend()
```

```python
# With Mean Centering

data = np.random.rand(100,2)
# Calculate the mean of the data of the along each dimension
data_mean = np.mean(data, axis=0)
print(data_mean)

# Perform mean centering
centered_data = data - data_mean

# Plot the mean_centered data
plt.subplot(1, 2, 2)
plt.scatter(centered_data[:, 0], centered_data[:, 1], label="Mean centered data", color="orange")
plt.xlabel("X_label")
plt.ylabel("Y_label")
plt.title("After Mean centering")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend()
plt.show()
```

## Dot Product

```python
# Define two vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
print(np.dot(A, B))
print(A@B)
```

```python
# Define three Vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
C = np.array([7, 8, 9])

# Calculate dot product using numpy.dot() function
dot_product_AB = np.dot(A, B)
dot_product_BA = np.dot(B, A)
print(dot_product_AB)
print(dot_product_BA)
```

```python
# Check commutative property : A.B = B.A

if dot_product_AB == dot_product_BA:
    print("The dot product is commutative.")

else:
    print("The dot product is not commutative.")
```

```python
# Check distribution property : A.B(B + C) = A.B + A.C
dot_product_A_BC = np.dot(A, B + C)
dot_product_AB_plus_AC = np.dot(A,B) + np.dot(A,C)

print(dot_product_A_BC)
print(dot_product_AB_plus_AC)

if dot_product_A_BC == dot_product_AB_plus_AC:
    print("The dot product is distributive.")
else:
    print("The dot product is not distributive.")
```

## Angle Between 2 Vectors

```python
# Define two vectors
A = np.array([1, 2, 3])
B = np.array([-4, -5, -6])
C = np.array([5, 5, 5])

# Calculate the cosine similarity A.B
cosine_similarity1 = np.dot(A,B) / (np.linalg.norm(A) * np.linalg.norm(B))
print("Cosine similarity between A and B :", cosine_similarity1)

# Caculate the cosine similarity A.C
cosine_similarity2 = np.dot(A,C) / (np.linalg.norm(A) * np.linalg.norm(C))
print("Cosine similarity bteween A and C :", cosine_similarity2)
```
