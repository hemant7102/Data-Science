```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Colored Scatter Plot

```python
iris = pd.read_csv('IRIS.csv')
```

```python
iris
```

```python
# Convert species into number which will help faster to fetch

iris['species'] = iris['species'].replace({'Iris-setosa':0, 'Iris-virginica':1, 'Iris-versicolor':2})

```

```python
# By default viridis
plt.scatter(iris['sepal_length'], iris['petal_length'], c=iris['species'], cmap='summer') # cmap --> summer, winter, jet, viridis

plt.xlabel('sepal_length')
plt.ylabel('petal_length')

plt.legend()

plt.colorbar()
```

# Annotation

```python
batter = pd.read_csv('batter.csv')
```

```python
batter
```

```python
sample_df = batter.head(100).sample(25, random_state = 5)
```

```python
sample_df.head()
```

```python
sample_df.shape
```

```python
plt.figure(figsize=(10,10))

plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
plt.figure(figsize=(10,10))

plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])

plt.axhline(130, color='red')

plt.axvline(30, color='red')
```

### 3> Surface PLot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = xx**2 + yy**2

fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

# Contour Plots  (2D Plots)

```python

fig = plt.figure(figsize = (12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define grid
x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 50)

xx, yy = np.meshgrid(x, y)

# Define a function for z
z = np.sin(xx) + np.cos(yy)

# Plot
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()

p = ax.contour(xx, yy, z, cmap='viridis')
fig.colorbar(p)
plt.show()
```

# 5> HeatMap

```python
delivery = pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run'] == 6)]
temp_df
```

```python
grid = temp_df.pivot_table(index='overs', columns='ballnumber', values='batsman_run', aggfunc='count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(), list(range)(1,21))
plt.xticks(np.arange(0,6), list(range)(1,7))

plt.colorbar()
```
