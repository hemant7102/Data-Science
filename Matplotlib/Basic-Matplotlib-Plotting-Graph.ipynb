```python
batsman.head()
```

### PLotting Multiple Plots

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'])
```

### Labels title

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

### Color(hex) and Line(width and style)

```python
# color

plt.plot(batsman['index'], batsman['V Kohli'], color='#D9F10F')   # here we use hex code 

plt.plot(batsman['index'], batsman['RG Sharma'], color='grey')  # here we customize by own choise color

plt.title('Rohit Sharma Vs Virat Kohli career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='grey', linestyle='dashed', linewidth=3)

plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='solid', linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='grey', linestyle='dashed', linewidth=3, markersize=3)

plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='solid', linewidth=2, markersize=6)

plt.title('Rohit Sharma Vs Virat Kohli Career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

### Legend  --> Location

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='grey', linestyle='dashed', linewidth=3)

plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='solid', linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend()
```

### Limiting axes

```python
price = [48000,54000,57000,49000,45000,47000,450000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0, 100000)
```

```python
price = [48000,54000,57000,49000,45000,47000,450000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0, 100000)
plt.xlim(2017,2019)
```

### Grid

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='grey', linestyle='dashed', linewidth=3)

plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='solid', linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.grid()
plt.legend()
```

### Show()

```python
# Show is used plot garph whenever we plot graph but not visible then.

plt.plot(batsman['index'], batsman['V Kohli'], color='grey', linestyle='dashed', linewidth=3)

plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='solid', linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Camparison')

plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.grid()
plt.legend()
plt.show()
```

### Dataset

```python
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'])
plt.show()
```

```python
# TO show percentage there is function called (autopct=%0.1f%%)

plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

### Colors

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', colors=['blue','green','yellow','pink','cyan', 'brown'])
plt.show()
```
