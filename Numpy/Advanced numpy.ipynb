```python
## Numpy array  VS  Python lists

# 1> Speed of List

import time

a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]
c = []
start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])

print(time.time() - start)    
```

```python
# 2> Numpy

import numpy as np

a = np.arange(10000000)
b = np.arange(10000000,20000000)
start = time.time()
c = a + b
print(time.time() - start)
```

```python
## Broadcasting 

# 1> Same shape

a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print(a)
print(b)

print(a + b)
```

```python
# 2> Different Shape

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a)
print(b)
print(a + b)
```

```python
## 1> Sigmoid  ---> Machine Learninig and Deep Learning ----> Logistic Regression

def sigmoid(array):
    return 1 / (1 + np.exp(-(array)))     # Formula of Sigmoid    S(x) = 1 / 1 + e ^-x

a = np.arange(100)
sigmoid(a)
```

```python
# 2> Mean Squared error  ----> Machine Learninig ----> Loss Function

actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)     # Formula for mean squared error   (yi - y^i)^2  1> yi --> observed value
actual                                                                                       # y^i --> predicted value
```

```python
predicted
```

```python
def mse(actual,predicted):
    return np.mean((actual - predicted)**2)

mse(actual,predicted)    
```

```python
# 3> Binary Cross / Log Loss  --->   -(yi * log(pi)  + (1 - yi) * log(1 - pi))

actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
actual
```

```python
predicted
```

```python
import numpy as np

def bce(actual, predicted):
    actual = np.array(actual)
    predicted = np.array(predicted)
    
    eps = 1e-15
    predicted = np.clip(predicted, eps, 1 - eps)
    
    loss = -(actual * np.log(predicted) + (1 - actual) * np.log(1 - predicted))
    return np.mean(loss) 


actual = [1, 50,25]
predicted = [1,50, 25]

print(bce(actual, predicted))
    
```
