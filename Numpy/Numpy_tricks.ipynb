```python
## 17> Set Function


m = np.array([1,2,3,4,5])
n = np.array([3,4,5,6,7])
```

```python
# 1> np.union1d

np.union1d(m,n)   # all value present in both array
```

```python
# 2> np.intersect1d

np.intersect1d(m,n)   # The value which are comman in both array
```

```python
# 3> np.setdiff1d

np.setdiff1d(m,n) # the value not present in n but present in m
```

```python
np.setdiff1d(n,m)
```

```python
# 4> np.setxor1d

np.setxor1d(m,n)  # Not comman in both array
```
